# N011 · 时间与空间复杂度

**目标：** 能按输入规模分析操作数和额外空间。

**先修：** N007, N010。本章自包含，依次运行即可；建议先读推导，再预测输出。

**知识点：** 渐进上界；最坏/平均；递归栈；输出空间；位复杂度提醒。

## 从问题到算法

复杂度描述输入规模变化时操作数如何增长，不是给某台机器测一组毫秒数。两数和暴力法枚举 i<j，从而不漏任何不同索引对，也不重复枚举交换次序。外层各次内层长度分别为 n−1、n−2、…、0。

$$
\sum_{i=0}^{n-1}(n-1-i)=\frac{n(n-1)}2.
$$

## 最小实现

**本章接口：** `two_sum_bruteforce(nums, target)`、`count_pair_checks(n)`

```python
def two_sum_bruteforce(nums,target):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return [i,j]
    return None

def count_pair_checks(n):
    return n*(n-1)//2
```

## 正确性、前提与复杂度

任何两个不同索引经排序后唯一写成 i<j，因此双循环完备且不重复。找到即返回会缩短某些输入的时间，但不改变最坏二次界。

**代价：** 最坏 O(n²) 时间，O(1) 辅助空间，不计常数大小答案。
